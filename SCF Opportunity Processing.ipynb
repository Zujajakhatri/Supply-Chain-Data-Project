# 📊 Dataset Overview: Supply Chain Finance Opportunity Data

## 1. Description

This project evaluates **Supply Chain Finance (SCF) opportunities across Pakistani districts** by integrating SME, supply-chain, infrastructure, banking, financing etc.

The analysis identifies districts and dominant industries with **higher SCF potential**, assesses financing needs, and maps suitable banking products to local supply-chain characteristics.

The final dataset supports **district-level opportunity analysis and Power BI visualization** for strategic banking decisions.


---

## 2. Dataset Structure & Key Dimensions

The analysis uses **three worksheets** containing complementary district-level information:

| **Sheet**   | **Focus Area**                      | **Key Information**                                                                                       |

| **Sheet 1** | Supply Chain Data     | SME activity, anchor sectors, employment, exports, banking penetration, value chains                      |

| **Sheet 3** | Import & Export Data            | Major export products, destinations, import products and import sources                                   |

# Feature Engineering

Feature engineering was performed to transform qualitative variables into numerical values and create new analytical features. This step prepares the dataset for scoring, comparison, and further analysis.

## SME Scale Score

Classify districts into **three SME scale levels** based on the estimated number of SMEs. This scoring helps identify districts with a **relatively larger SME base**, which may indicate greater potential for supply chain financing.


```python
# SME Scale Score
df_final["SME_Scale_Score"] = pd.qcut(
    df_final["Estimated_Number_of_SMEs"].rank(method="first"),
    q=3,
    labels=[1, 2, 3]
).astype(int)

df_final[["District", "Estimated_Number_of_SMEs", "SME_Scale_Score"]].head()
```

### Outcome

Each district is assigned an **SME Scale Score from 1 to 3**, representing its relative SME concentration and supporting the assessment of **district-level SCF potential**.


## Employment Scale Score

Classify districts into **three employment scale levels** based on estimated SME employment. This helps identify districts with a **larger workforce base**, indicating greater economic and potential financing activity.


```python
# Employment Scale Score
df_final["Employment_Scale_Score"] = pd.qcut(
    df_final["Estimated_Number_of_Employees"].rank(method="first"),
    q=3,
    labels=[1, 2, 3]
).astype(int)

df_final[[
    "District",
    "Estimated_Number_of_Employees",
    "Employment_Scale_Score"
]].head()
```

### Outcome

Each district is assigned an **Employment Scale Score from 1 to 3**, reflecting its relative SME employment size.


## Anchor Firm Score

Convert anchor-firm presence into a **binary score** to identify districts connected to established businesses that can support structured supply-chain financing.


```python
# Anchor Firm Score
df_final["Anchor_Firm_Score"] = (
    df_final["Anchor_firm_presence"]
    .astype(str)
    .str.strip()
    .str.lower()
    .map({
        "yes": 1,
        "no": 0
    })
    .fillna(0)
)

df_final[["District", "Anchor_firm_presence", "Anchor_Firm_Score"]].head()
```

### Outcome

Each district is assigned an **Anchor Firm Score of 1 or 0**, based on the presence or absence of an anchor firm.


## Production Capacity Score

Convert production volume into a **three-level score** to measure the scale of local production. Higher production capacity indicates **greater supply-chain activity and potential financing demand**.


```python
# Production Capacity Score
df_final["Production_Capacity_Score"] = (
    df_final["Production_Volume"]
    .astype(str)
    .str.strip()
    .str.lower()
    .map({
        "high": 3,
        "medium": 2,
        "low": 1
    })
    .fillna(1)
)

df_final[[
    "District",
    "Production_Volume",
    "Production_Capacity_Score"
]].head()
```

### Outcome

Each district is assigned a **Production Capacity Score from 1 to 3**, based on its production volume level.

Convert processing capacity into a **three-level score** to measure the district's ability to process local production and support supply-chain activity.

## Business Environment Score

Convert key business-environment indicators into **three-level scores** and combine them into an overall score to assess the district's **business and operational readiness for SCF**.


```python
# Convert business environment indicators to scores

business_cols = [
    "Business_Registration",
    "Tax_System",
    "Property_Registration",
    "Contract_Enforcement",
    "Legal & Regulatory_Environment",
    "Skilled_Labor_Availability",
    "Export & Trade_Facilitation"
]

score_map = {
    "high": 3,
    "medium": 2,
    "low": 1
}

for col in business_cols:
    df_final[col + "_Score"] = (
        df_final[col]
        .astype(str)
        .str.strip()
        .str.lower()
        .map(score_map)
        .fillna(1)
    )

df_final["Business_Environment_Score"] = (
    df_final[[col + "_Score" for col in business_cols]]
    .mean(axis=1)
    .round(2)
)

df_final[[
    "District",
    "Business_Environment_Score"
]].head()
```

### Outcome

A **Business Environment Score** is calculated for each district by averaging the selected indicators. Higher scores indicate a more supportive business environment.


## Supply Chain Potential Score

Combine SME scale, employment, anchor firms, production, processing capacity, and logistics to measure the district's **overall supply-chain potential**.


```python
# Supply Chain Potential Score

supply_chain_cols = [
    "SME_Scale_Score",
    "Employment_Scale_Score",
    "Anchor_Firm_Score",
    "Production_Capacity_Score",
    "Processing_Capacity_Score",
    "Logistics_Score"
]

df_final["Supply_Chain_Potential_Score"] = (
    df_final[supply_chain_cols]
    .mean(axis=1)
    .round(2)
)

df_final[[
    "District",
    "Supply_Chain_Potential_Score"
]].head()
```

### Outcome

A **Supply Chain Potential Score** is calculated for each district, with higher scores indicating stronger supply-chain potential.


## Financing Need Score

Combine credit access, banking access, working capital gap, and supply-chain constraints to measure the district's **overall financing need**.


```python
# Financing Need Score

financing_cols = [
    "Credit_Access_Score",
    "Banking_Access_Score",
    "Working_Capital_Gap_Score",
    "Supply_Chain_Constraint_Score"
]

df_final["Financing_Need_Score"] = (
    df_final[financing_cols]
    .mean(axis=1)
    .round(2)
)

df_final[[
    "District",
    "Financing_Need_Score"
]].head()
```

### Outcome

A **Financing Need Score** is calculated for each district, with higher scores indicating **greater potential financing needs**.


## Digital Readiness Score

Combine digital payment adoption, infrastructure, and business environment indicators to measure the district's **overall digital and operational readiness for SCF**.


```python
# Digital Readiness Score

df_final["Digital_Readiness_Score"] = (
    df_final[
        [
            "Digital_Payment_Score",
            "Infrastructure_Score",
            "Business_Environment_Score"
        ]
    ]
    .mean(axis=1)
    .round(2)
)

df_final[[
    "District",
    "Digital_Readiness_Score"
]].head()
```

### Outcome

A **Digital Readiness Score** is calculated for each district, with higher scores indicating greater readiness for digitally enabled SCF.


## Trade Finance Potential Score

Combine export orientation, import activity, and trade activity indicators to measure the district's **potential demand for trade finance within its supply chains**.


```python
# Trade Finance Potential Score

trade_cols = [
    "Export_Potential_Score",
    "Import_Potential_Score",
    "Export_Activity_Score",
    "Import_Activity_Score"
]

df_final["Trade_Finance_Potential_Score"] = (
    df_final[trade_cols]
    .mean(axis=1)
    .round(2)
)

df_final[[
    "District",
    "Trade_Finance_Potential_Score"
]].head()
```

### Outcome

A **Trade Finance Potential Score** is calculated for each district, with higher scores indicating greater trade-related financing potential.


## Final SCF Opportunity Score

Combine the key opportunity components using **weighted scoring** and normalize the result to a **0–100 scale** to create a single district-level SCF opportunity measure.


```python
# Final Supply Chain Financing Opportunity Score

df_final["Supply_Chain_Financing_Opportunity_Score"] = (
    df_final["Supply_Chain_Potential_Score"] * 0.25 +
    df_final["Financing_Need_Score"] * 0.25 +
    df_final["Trade_Finance_Potential_Score"] * 0.20 +
    df_final["Digital_Readiness_Score"] * 0.15 +
    df_final["Export & Trade_Facilitation_Score"] * 0.15
) / 3 * 100

df_final["Supply_Chain_Financing_Opportunity_Score"] = (
    df_final["Supply_Chain_Financing_Opportunity_Score"]
    .round(2)
)
df_final[[
    "District",
    "Supply_Chain_Financing_Opportunity_Score"
]].sort_values(
    "Supply_Chain_Financing_Opportunity_Score",
    ascending=False
).head(10)
```

### Outcome

Each district receives a **Supply Chain Financing Opportunity Score from 0 to 100**, allowing districts to be compared based on their overall SCF potential.


## Opportunity Level

Classify districts into **Low, Medium, and High opportunity levels** based on their final SCF Opportunity Score to simplify district-level comparison.


```python
# Opportunity Level

df_final["Opportunity_Level"] = pd.cut(
    df_final["Supply_Chain_Financing_Opportunity_Score"],
    bins=[0, 60, 80, 100],
    labels=["Low", "Medium", "High"],
    include_lowest=True
)

df_final[[
    "District",
    "Supply_Chain_Financing_Opportunity_Score",
    "Opportunity_Level"
]].head()
```

### Outcome

Each district is classified into **Low, Medium, or High SCF opportunity** based on its score range.

## Recommended Banking Product

SCF products are recommended using a **rule-based approach** that considers industry, value chain, anchor-firm presence, export orientation, banking penetration, and SME opportunity.

### Distributor / Channel Finance

Recommended for **retail, wholesale, and trading ecosystems** where financing is required across distributors, dealers, and sales channels.

### PO Finance

Recommended for **confirmed purchase-order based businesses**, particularly export-oriented manufacturers and suppliers requiring funds to fulfill orders.

### Payables Finance

Recommended where **anchor firms and established buyer–supplier relationships** exist, enabling financing of supplier invoices linked to anchor buyers.

### Receivables Discounting / Factoring

Recommended for businesses with **regular receivables and cash-flow cycles**, such as transport, logistics, professional services, and hospitality.

### Inventory Finance

Recommended for **inventory-intensive businesses** requiring working capital to purchase, hold, or process stock before sale.

### Debt-Based Financing

Recommended for **mining and mineral-related businesses** where financing requirements are more closely linked to capital-intensive operations.


```python
def recommend_scf_product(row):

    sector = str(row["Key_Anchor_Sectors"]).lower()
    value_chain = str(row["Dominant_Value_Chain"]).lower()
    
    anchor = str(row["Anchor_firm_presence"]).lower() == "yes"
    export = str(row["Export_Oriented"]).lower() == "yes"
    
    banking = row["Banking_Penetration (%)"]
    opportunity = row["SME Opportunity_Score"]

    text = sector + " " + value_chain

    # 1. Distributor / Channel Finance
    if any(x in text for x in [
        "retail", "wholesale", "trade"
    ]):
        return "Distributor / Channel Finance"

    # 2. Manufacturing / Industrial Supply Chains
    if any(x in text for x in [
        "manufacturing", "textile", "apparel",
        "light engineering", "tobacco",
        "chemicals", "plastics"
    ]):
        
        if export and anchor:
            return "PO Finance + Payables Finance + Receivables Discounting"
        
        elif anchor:
            return "Payables Finance + Inventory Finance"
        
        else:
            return "Inventory Finance"

    # 3. Agriculture / Agri Supply Chains
    if any(x in text for x in [
        "agriculture", "agribusiness", "horticulture",
        "fruit", "rice", "sugarcane",
        "date palm", "livestock", "dairy",
        "fisheries"
    ]):
        
        if export:
            return "PO Finance + Inventory Finance"
        
        else:
            return "Inventory Finance"

    # 4. Mining / Minerals
    if any(x in text for x in [
        "mining", "minerals"
    ]):
        return "Debt‑Based Financing"

    # 5. Transport / Logistics
    if any(x in text for x in [
        "transport", "logistics"
    ]):
        return "Receivables Discounting / Factoring, PO Finance"

    # 6. IT / Technology Startups
    if any(x in sector for x in [
        "IT & Services",
        "Information Technology (IT)"
        
    ]):
        return "PO Finance"


    # 7. Professional Services / Hospitality / Tourism
    if any(x in text for x in [
        "professional services",
        "hospitality",
        "tourism"
    ]):
        return "Receivables Discounting / Factoring"


    # 8. General anchor-firm ecosystem
    if anchor and banking >= 70 and opportunity >= 30:
        return "Payables Finance"


    # 9. Default
        return "Inventory Finance"
        
df_final["Recommended_Banking_Product"] = df_final.apply(
    recommend_scf_product,
    axis=1
)
```
